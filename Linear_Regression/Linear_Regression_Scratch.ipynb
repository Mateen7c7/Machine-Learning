```python
import numpy as np
import matplotlib.pyplot as plt


x = np.arange(1,30)
y = 2*x + 3
```

```python
w_in = 0
b_in = 0
```

```python
def cost_function(x,y,w,b):
    f_x = (w*x)+b
    j = np.sum((f_x - y)**2)*1/(2*len(x))
    return j

# cost_function(x,y,2,3)
```

```python
def gradients(x,y,w,b):
    dj_dw = np.sum(x*(((w*x)+b)-y))
    dj_db = np.sum((((w*x)+b)-y))

    dj_db = dj_db / len(x)
    dj_dw = dj_dw / len(x)

    return dj_dw,dj_db

# g,h = gradients(x,y,w,b)
# print(g,h)
```

```python
def gradient_descent(x,y,w,b,n,a):
    for i in range(n):
        dj_dw ,dj_db = gradients(x,y,w,b)
        w = w - (a * dj_dw)
        b = b - (a * dj_db)
        if (i + 1) % max(1, n // 10) == 0 and i < 10 * max(1, n // 10):
            print(f"Iteration {i+1}: w = {w}, b = {b}")
    return w , b
```

```python
a = 0.001
n = 25000
```

```python
w,b = gradient_descent(x,y,w_in,b_in,n,a)
print(f"Final  w = {w} , b = {b}")
```
