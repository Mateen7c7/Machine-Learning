```python
# Import numpy 
import numpy as np

x = np.arange(-2,10)

# Cost Function
j = (x - 5)**2

dj_dx = 2*(x-5)

matrix = np.column_stack((x,j,dj_dx))
print("    x   j   dj/dx")
print(matrix)
```

```python
x = np.arange(1,10)

y = 2*x

w = np.arange(-3,7)

# print(x,y)

m = len(x)

def cost(w):
    j = np.sum((w*x - y)**2) * 1 / (2*m)
    return j

def diff(w):
    dj = np.sum((x)*((w*x)-y))*1/m
    return dj

print(f"{'w':>5} {'j':>10} {'dj_dw':>10}")
print("-" * 28)

for i in range(len(w)):
    print(f"{w[i]:>5} {cost(w[i]):>10.2f} {diff(w[i]):>10.2f}")
```

So we can see here how cost (j) and dj/dw especially dj/dj tells us how to adjust w to get the right value for w to minimize j(x)

```python
x = np.arange(1,10)

y = 2*x+3

m = len(x)

w = np.arange(0,4)
b = np.arange(2,5)

def cost(w,b):
    j = np.sum((((w*x)+b)-y)**2)*1/(2*m)
    return j

def dj_dw(w,b):
    res = np.sum(x*(((w*x)+b)-y))*1/m
    return res

def dj_db(w,b):
    res = np.sum((((w*x)+b)-y))*1/m
    return res

print(f"{'w':>5} {'b':>5} {'j':>10} {'dj_dw':>10} {'dj_db':>10}")
print("-" * 46)

for i in w:
    for k in b:
        print(f"{i:>5} {k:>5} {cost(i,k):>10.2f} {dj_dw(i,k):>10.2f} {dj_db(i,k):>10.2f}")
```
